```python
#label encoding
import pandas as pd
#scikit-learn is a library
#preprocessing is a module
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
#labelencoder is a class has already inbuilt methods and members
```

```python
df=pd.DataFrame({
    'Product':['Watch','Mouse','stove','shoe','stove','Mouse','Watch','shoe'],
    'subjects':['sql','Sql','Testing','genai','Genai','Testing','Testing','genai'],
    'objects':['marker','Duster','Duster','Marker','Pen','Duster','Pencil','Pen']

})
df
#DataFrame-understands only dict-constructor that creates dataframe
```

```python
df
```

```python
df['Product'].unique()
df['subjects'].unique()
df['objects'].unique()
```

```python
le= LabelEncoder()
```

```python
#fit_transform- METHOD IN LABELENCODER
#transform- transforms mapped data into output
#.fit- will identify unique values-sorts amd maps

df['Product']= le.fit_transform(df['Product'])
df['subjects']=le.fit_transform(df['subjects'])
df['objects']=le.fit_transform(df['objects'])
df
#overwriting in original column
```

```python
#OneHot Coder
df1=pd.DataFrame({
    'Animals':['Lion','Mouse','Sheep','Sheep','Tiger','Mouse','Donkey','Sheep']})
df1
```

```python
ohe= OneHotEncoder(sparse_output=False)
```

```python
data= ohe.fit_transform(df1[['Animals']])
name=ohe.get_feature_names_out(['Animals'])
```

```python
data
name
```

```python

df1= pd.DataFrame(data, columns=name)
```

```python
df1
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df2=pd.DataFrame({'Shapes':['circle','square','rhombus','rectangle','square','circle']})
ohe=OneHotEncoder(sparse_output=False)
data1=ohe.fit_transform(df2[['Shapes']])
name1=ohe.get_feature_names_out(['Shapes'])
df2=pd.DataFrame(data1,columns=name1)
df2
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df3=pd.DataFrame({'directions':['east','west','north','south','east','west']})
ohe=OneHotEncoder(sparse_output=False)
data2=ohe.fit_transform(df3[['directions']])
name2=ohe.get_feature_names_out(['directions'])
df3=pd.DataFrame(data2,columns=name2)
df3
```
