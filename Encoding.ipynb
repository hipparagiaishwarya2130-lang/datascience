```python
#label encoding
import pandas as pd
#scikit-learn is a library
#preprocessing is a module
from sklearn.preprocessing import LabelEncoder
#labelencoder is a class has already inbuilt methods and members
```

```python
df=pd.DataFrame({
    'Product':['Watch','Mouse','stove','shoe','stove','Mouse','Watch','shoe'],
    'subjects':['sql','Sql','Testing','genai','Genai','Testing','Testing','genai'],
    'objects':['marker','Duster','Duster','Marker','Pen','Duster','Pencil','Pen']

})
df
#DataFrame-understands only dict-constructor that creates dataframe
```

```python
df
```

```python
df['Product'].unique()
df['subjects'].unique()
df['objects'].unique()
```

```python
le= LabelEncoder()
```

```python
#fit_transform- METHOD IN LABELENCODER
#transform- transforms mapped data into output
#.fit- will identify unique values-sorts amd maps

df['Product']= le.fit_transform(df['Product'])
df['subjects']=le.fit_transform(df['subjects'])
df['objects']=le.fit_transform(df['objects'])
df
#overwriting in original column
```
