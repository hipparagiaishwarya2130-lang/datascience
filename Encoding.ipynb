```python
#label encoding
import pandas as pd
#scikit-learn is a library
#preprocessing is a module
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
#labelencoder is a class has already inbuilt methods and members
```

```python
df=pd.DataFrame({
    'Product':['Watch','Mouse','stove','shoe','stove','Mouse','Watch','shoe'],
    'subjects':['sql','Sql','Testing','genai','Genai','Testing','Testing','genai'],
    'objects':['marker','Duster','Duster','Marker','Pen','Duster','Pencil','Pen']

})
df
#DataFrame-understands only dict-constructor that creates dataframe
```

```python
df
```

```python
df['Product'].unique()
df['subjects'].unique()
df['objects'].unique()
```

```python
le= LabelEncoder()
```

```python
#fit_transform- METHOD IN LABELENCODER
#transform- transforms mapped data into output
#.fit- will identify unique values-sorts amd maps

df['Product']= le.fit_transform(df['Product'])
df['subjects']=le.fit_transform(df['subjects'])
df['objects']=le.fit_transform(df['objects'])
df
#overwriting in original column
```

```python
#OneHot Coder
df1=pd.DataFrame({
    'Animals':['Lion','Mouse','Sheep','Sheep','Tiger','Mouse','Donkey','Sheep']})
df1
```

```python
ohe= OneHotEncoder(sparse_output=False)
```

```python
data= ohe.fit_transform(df1[['Animals']])
name=ohe.get_feature_names_out(['Animals'])
```

```python
data
name
```

```python

df1= pd.DataFrame(data, columns=name)
```

```python
df1
```

```python
#ordinal encoding-single column
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
df5=pd.DataFrame({'places':['rome','paris','turkey','rome','veitnam']})
oe= OrdinalEncoder(categories=[['paris','rome','turkey','veitnam']])
df5['places']= oe.fit_transform(df5[['places']])
df5
```

```python
#ordinal encoding-multiple column
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df6=pd.DataFrame({'sizes':['s','m','l','s'],'countries':['india','china','europe','india']})
oe= OrdinalEncoder(categories=[['s','m','l'],['europe','china','india']])
df6[['sizes','countries']]= oe.fit_transform(df6[['sizes','countries']])
df6
```

```python
#onehotencoding-multiple columns
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df7=pd.DataFrame({'sizes':['s','m','l','s'],'countries':['india','china','europe','india']})
ohe=OneHotEncoder(sparse_output=False)
dataa= ohe.fit_transform(df7[['sizes','countries']])
namee=ohe.get_feature_names_out(['sizes','countries'])
df7=pd.DataFrame(dataa,columns=namee)
df7
```
