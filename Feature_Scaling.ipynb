FEATURE SCALING-

```python
#Standardisation
import pandas as pd
from sklearn.preprocessing import StandardScaler
df=pd.DataFrame({'grams':[2,3,4,5],'price':[200,500,800,1200]})
ss=StandardScaler()
#fit-learning data,calculating mean,devitaion done by fit
#transform-tranforming or reflecting it to original data
df_scaled=ss.fit_transform(df)
df_scaled
```

```python
df2=pd.DataFrame(df_scaled,columns=df.columns)
df2
```

```python
#Standardisation
import pandas as pd
from sklearn.preprocessing import StandardScaler
df1=pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/hr_data.csv')
ss1=StandardScaler()
n= df1.select_dtypes(include=['number']).columns
df1_numeric=df1[n]
df_scaled1=ss1.fit_transform(df1_numeric)
df_scaled1=pd.DataFrame(df_scaled1,columns=df1_numeric.columns)
df_scaled1
```

```python
#minmax scaler
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df1=pd.DataFrame({'height':[160,154,148],'weight':[47,52,62]})
mm=MinMaxScaler()
M=mm.fit_transform(df1)
M=pd.DataFrame(M,columns=df1.columns)
M
```

```python
#minmax scaler
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df2=pd.DataFrame({'score_py':[144,113,122],'score_sql':[66,89,45]})
mm1=MinMaxScaler()
M=mm1.fit_transform(df2)
M=pd.DataFrame(M,columns=df2.columns)
M
```

```python
#normalizer
import pandas as pd
from sklearn.preprocessing import Normalizer
df3=pd.DataFrame({'height':[160,154,148],'weight':[47,52,62]})
mm2=Normalizer()
M=mm2.fit_transform(df3)
M=pd.DataFrame(M,columns=df3.columns)
M
```
